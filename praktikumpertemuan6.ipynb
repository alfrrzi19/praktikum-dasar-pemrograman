```python
angka = [10, 20, 30, 40, 50, 60]

print("Indeks 1 sampai sebelum indeks 4:", angka[1:4])
print("Tiga elemen pertama:", angka[:3])
print("Elemen mulai indeks 3 sampai akhir:", angka[3:])
print("List terbalik:", angka[::-1])
```

```python
daftar_matkul = ["Algoritma", "Tipe Data"]
daftar_matkul.append("Basis Data")
print(daftar_matkul)
daftar_matkul.insert(1, "Pemrograman")
print(daftar_matkul)
del daftar_matkul[0]
print(daftar_matkul)
terhapus = daftar_matkul.pop()
print("Terhapus:", terhapus)
print(daftar_matkul)
```

```python
nilai = [85, 70, 92, 60, 78]

print("Jumlah data:", len(nilai))
print("Total:", sum(nilai))
print("Rata-rata:", sum(nilai) / len(nilai))
print("Urut naik:", sorted(nilai))
print("Urut turun:", sorted(nilai, reverse=True))
print("List asli:", nilai)
```
