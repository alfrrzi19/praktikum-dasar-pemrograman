```python
def buat_biodata(nama, umur, jurusan):
    print("Nama:", nama)
    print("Umur:", umur)
    print("Jurusan:", jurusan)
    print()

# Positional argument
buat_biodata("Alfa", 18, "Sistem Informasi")

# Keyword argument dengan urutan berbeda
buat_biodata(jurusan="Informatika", nama="Budi", umur=20)
```

```python
def konversi_menit_ke_detik(menit):
    """
    Mengonversi menit menjadi detik.
    Parameter:
        menit: jumlah waktu dalam menit.
    Return:
        Jumlah waktu dalam detik.
    """
    detik = menit * 60
    return detik

print(konversi_menit_ke_detik(3))
print(konversi_menit_ke_detik.__doc__)
```

```python
def hitung_subtotal(harga, jumlah):
    return harga * jumlah

def hitung_diskon(subtotal):
    if subtotal >= 100000:
        return subtotal * 0.10
    else:
        return 0

def tampilkan_struk(nama_barang, subtotal, diskon):
    total_akhir = subtotal - diskon

    print("===== STRUK PEMBELIAN =====")
    print("Nama Barang:", nama_barang)
    print("Subtotal: Rp", int(subtotal))
    print("Diskon: Rp", int(diskon))
    print("Total Akhir: Rp", int(total_akhir))

subtotal = hitung_subtotal(25000, 5)
diskon = hitung_diskon(subtotal)

tampilkan_struk("Buku", subtotal, diskon)
```
