```python
panjang = float(input("Masukkan panjang: "))
lebar = float(input("Masukkan lebar: "))

luas = (panjang * lebar)
keliling = (2 * (panjang + lebar))

print("Luas", luas)
print("Keliling", keliling)
```

```python
fahrenheit = float (input("Masukkan suhu Fahrenheit: "))
celcius = (fahrenheit - 32) * 5/9
print(f"Suhu dalam Celcius: {celcius:.2f}°C")
```

```python
harga_satuan = float(input("Masukkan harga satuan: "))
jumlah_barang = int(input("Masukkan jumlah barang: "))
persentase_diskon = float(input("Masukkan persentase diskon (contoh: 3 untuk 3%): "))

subtotal = harga_satuan * jumlah_barang
nominal_diskon = subtotal * (persentase_diskon / 100)
total_bayar = subtotal - nominal_diskon

print(f"\nSubtotal: Rp{subtotal:,.2f}")
print(f"Nominal Diskon ({persentase_diskon}%): Rp{nominal_diskon:,.2f}")
print(f"Total Bayar: Rp{total_bayar:,.2f}")
```

```python
total_detik = int(input("Masukkan total detik: "))

jam = total_detik // 3600
sisa_detik = total_detik % 3600
menit = sisa_detik // 60
detik = sisa_detik % 60

print(f"{jam:02d}:{menit:02d}:{detik:02d}")
```
